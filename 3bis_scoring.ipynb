# Notebook 3bis — Scoring rigoureux

## Objectif

Calculer le **coût de routage** propre pour chaque arête IdF, en s'appuyant sur :
- La littérature route choice (Broach 2012, Menghini 2010)
- Les profils production BRouter / GraphHopper
- Les règles métier spécifiques du club de route parisien

On construit une **fonction de coût multiplicative** à trois composantes :

$$\text{coût}(e) = L(e) \cdot f_{\text{infra}}(e, \text{profil}) \cdot f_{\text{pente}}(e, \text{profil}) \cdot f_{\text{plaisir}}(e, \text{profil}, \alpha)$$

Avec une logique hiérarchique :
1. **Niveau 1 — exclusion** : certaines arêtes sont retirées du graphe (coût infini)
2. **Niveau 2 — pénalités dures** : facteurs × 10 à × 20
3. **Niveau 3 — pondération fine** : facteurs × 0.8 à × 3

## Papiers de référence

- **Broach, Dill, Gliebe (2012)** *Transp. Res. Part A* 46(10):1730-1740 — coefficients de référence en "équivalents distance" sur 164 cyclistes GPS de Portland
- **Menghini et al. (2010)** *Transp. Res. Part A* 44(9):754-765 — pente max > pente moyenne pour Zurich
- **Hood, Sall, Charlton (2011)** *Transportation Letters* 3(1):63-75 — cyclistes expérimentés ≠ casual
- **Meister et al. (2024)** *J. Cycling and Micromobility Research* — review récente

## Règles métier spécifiques (vélo de route club parisien)

- **Exclusion dure** des `trunk`, `motorway_link`, des `primary`/`secondary` à `maxspeed ≥ 90`
- **Exclusion dure** aussi des `primary` sans info maxspeed (conservateur)
- **Pas de pénalité forte sur les grosses pentes** (le club cherche les côtes)
- **Pistes cyclables neutres** en profil club (ni bonus ni malus)
- **Résidentiel** : jugé selon `node_density` (village sympa vs ville chiante)
- **Zones vertes** : bonus appliqué à **tous** les profils
- **BNAC contextuel** : bonus uniquement sur axes dangereux

## 2. Exclusion des arêtes interdites

### La règle d'or

**On ne route JAMAIS** sur ces axes :
- `trunk`, `trunk_link`, `motorway`, `motorway_link` (voies rapides)
- `primary` ou `secondary` avec `maxspeed ≥ 90` (nationales type N14)
- `primary` sans info `maxspeed` (conservateur : on considère que c'est une nationale)
- `footway`, `pedestrian` avec `bicycle` = 'no' ou absent
- Toute arête avec `access = 'no'` ou `'private'`

### Implémentation

On ajoute une colonne `is_routable BOOLEAN` qui indique si l'arête est
utilisable. Le graphe du Notebook 4 fera `WHERE is_routable`. Pas besoin
de supprimer les arêtes, on préserve les données brutes.

````python
# Ajout colonne is_routable
with engine.begin() as conn:
    conn.execute(text("""
        ALTER TABLE osm_edges ADD COLUMN IF NOT EXISTS is_routable BOOLEAN DEFAULT TRUE;
    """))

# Application des exclusions
print("Application des exclusions...")
with engine.begin() as conn:
    # 1. Voies rapides : interdiction absolue
    n1 = conn.execute(text("""
        UPDATE osm_edges
        SET is_routable = FALSE
        WHERE highway IN ('trunk', 'trunk_link', 'motorway', 'motorway_link');
    """)).rowcount
    
    # 2. Primary/secondary rapides (nationales)
    n2 = conn.execute(text("""
        UPDATE osm_edges
        SET is_routable = FALSE
        WHERE highway IN ('primary', 'secondary')
          AND maxspeed IS NOT NULL
          AND maxspeed >= 90;
    """)).rowcount
    
    # 3. Primary sans info maxspeed (principe de précaution)
    n3 = conn.execute(text("""
        UPDATE osm_edges
        SET is_routable = FALSE
        WHERE highway = 'primary'
          AND maxspeed IS NULL;
    """)).rowcount
    
    # 4. Voies piétonnes sans bicycle explicite
    n4 = conn.execute(text("""
        UPDATE osm_edges
        SET is_routable = FALSE
        WHERE highway IN ('footway', 'pedestrian')
          AND (bicycle IS NULL OR bicycle != 'designated');
    """)).rowcount

    # 5. Accès interdit/privé
    n5 = conn.execute(text("""
        UPDATE osm_edges
        SET is_routable = FALSE
        WHERE highway = 'service' AND length_m > 500;
    """)).rowcount

print(f"Exclusions appliquées :")
print(f"  Voies rapides (trunk/motorway)       : {n1:,}")
print(f"  Primary/secondary >=90 km/h           : {n2:,}")
print(f"  Primary sans maxspeed (précaution)   : {n3:,}")
print(f"  Voies piétonnes non-vélo             : {n4:,}")
print(f"  Service long (parking/impasse)       : {n5:,}")

with engine.connect() as conn:
    total = conn.execute(text("SELECT COUNT(*) FROM osm_edges")).scalar()
    routable = conn.execute(text("SELECT COUNT(*) FROM osm_edges WHERE is_routable")).scalar()

pct = 100 * routable / total
print(f"\nArêtes routables : {routable:,} / {total:,} ({pct:.1f}%)")
````

## 3. Score club brut (heatmap normalisée)

Même logique qu'avant : on log-normalise le nombre de traces distinctes qui
empruntent chaque arête.

$$\text{score\_club}(e) = \min\!\left(1, \frac{\log(1 + n_e)}{\log(1 + n_{\max})}\right)$$

Ce score sera **remplacé** par `score_club_gen` au Notebook 5 (PU learning
qui généralise à toutes les arêtes). Pour maintenant, on se contente du brut.

````python
# Vérification que edge_scores existe
with engine.begin() as conn:
    conn.execute(text("""
        -- On détruit l'ancienne table pour forcer la mise à jour de la structure
        DROP TABLE IF EXISTS edge_scores CASCADE;
        
        -- On la crée avec toutes les bonnes colonnes
        CREATE TABLE edge_scores (
            edge_id BIGINT PRIMARY KEY REFERENCES osm_edges(edge_id) ON DELETE CASCADE,
            score_club   DOUBLE PRECISION DEFAULT 0,
            score_osm_club   DOUBLE PRECISION DEFAULT 0,
            score_osm_solo   DOUBLE PRECISION DEFAULT 0,
            score_strava DOUBLE PRECISION DEFAULT 0,
            cost_factor_club DOUBLE PRECISION DEFAULT 1,
            cost_factor_solo DOUBLE PRECISION DEFAULT 1,
            updated_at TIMESTAMP DEFAULT NOW()
        );
        
        INSERT INTO edge_scores (edge_id)
        SELECT edge_id FROM osm_edges
        ON CONFLICT (edge_id) DO NOTHING;
    """))

# score_club
with engine.begin() as conn:
    conn.execute(text("""
        WITH club AS (
            SELECT edge_id, COUNT(DISTINCT trace_id) AS n
            FROM trace_edges GROUP BY edge_id
        ),
        max_n AS (SELECT MAX(n) AS mx FROM club)
        UPDATE edge_scores es
        SET score_club = LEAST(1.0,
            LN(1 + c.n) / NULLIF(LN(1 + mx.mx), 0)
        )
        FROM club c, max_n mx
        WHERE es.edge_id = c.edge_id;
    """))

with engine.connect() as conn:
    stats = pd.read_sql(text("""
        SELECT
            ROUND(AVG(score_club) FILTER (WHERE score_club > 0)::numeric, 3) AS mean_club,
            ROUND(MAX(score_club)::numeric, 3) AS max_club
        FROM edge_scores;
    """), conn)

print(stats.to_string(index=False))
````

## 4. Fonction de coût multiplicative

### Structure

Pour chaque arête, on calcule un **cost_factor** ≥ 1 qui multiplie la distance
dans l'algorithme de Dijkstra :

$$\text{coût}(e) = L(e) \cdot \text{cost\_factor}(e)$$

Un `cost_factor = 1` signifie "arête idéale" (distance réelle).
Un `cost_factor = 2` signifie "arête deux fois moins intéressante" (Dijkstra
est prêt à faire 2× la distance ailleurs pour l'éviter).

On calcule **deux** versions : `cost_factor_club` et `cost_factor_solo`.

### Composants (ordre de calcul dans le SQL)

```
cost_factor = hierarchy_factor        (= 1 à 20 selon highway/maxspeed)
            × surface_factor           (= 1 à 4 selon surface)
            × bnac_factor              (contextuel, selon highway)
            × residential_factor       (selon node_density)
            × greenery_bonus           (<= 1 si proche de zones vertes)
            × cycleway_factor          (selon profil)
```

### Tableau de référence (profil club_road)

| highway | maxspeed | cost_factor hiérarchie |
|---|---|---|
| `tertiary`, `unclassified` | - | **1.0** (idéal) |
| `secondary` | < 70 | 1.3 |
| `secondary` | 70-80 | 2.5 |
| `primary` | < 70 | 2.0 |
| `primary` | 70-80 | 5.0 |
| `residential` | - | 1.2 (modulé par node_density) |
| `living_street` | - | 1.5 |
| `cycleway` | - | 1.0 (neutre en road) |
| `service` | - | 2.5 |
| `path` | bicycle=designated | 2.0 |
| Autres | - | varie |

### Profil solo_casual

Mêmes exclusions dures, mais coefficients plus doux sur les grosses routes
avec piste cyclable, et bonus sur les cycleway dédiées.

````python
# --- Calcul du cost_factor pour le profil CLUB_ROAD ---
print("Calcul cost_factor_club...")

with engine.begin() as conn:
    conn.execute(text("""
        WITH base AS (
            SELECT e.edge_id,
                -- 1. HIERARCHIE ROUTIERE (facteur multiplicatif)
                CASE
                    -- Références : petites routes de campagne
                    WHEN e.highway IN ('tertiary', 'unclassified') THEN 1.0
                    WHEN e.highway = 'tertiary_link' THEN 1.0
                    
                    -- Secondary : OK si lentes, malus si rapides
                    WHEN e.highway = 'secondary' AND (e.maxspeed IS NULL OR e.maxspeed < 70) THEN 1.3
                    WHEN e.highway = 'secondary' AND e.maxspeed BETWEEN 70 AND 89 THEN 2.5
                    WHEN e.highway = 'secondary_link' THEN 1.5
                    
                    -- Primary : pénalité mais laissé si lent (les primary >= 90 sont déjà exclus)
                    WHEN e.highway = 'primary' AND e.maxspeed < 70 THEN 2.0
                    WHEN e.highway = 'primary' AND e.maxspeed BETWEEN 70 AND 89 THEN 5.0
                    WHEN e.highway = 'primary_link' THEN 3.0
                    
                    -- Résidentiel : base neutre, modulé après par node_density
                    WHEN e.highway = 'residential' THEN 1.2
                    WHEN e.highway = 'living_street' THEN 1.5
                    
                    -- Cycleway : neutre en road
                    WHEN e.highway = 'cycleway' THEN 1.0
                    
                    -- Service : petites allées, souvent impasses
                    WHEN e.highway = 'service' THEN 2.5
                    
                    -- Path avec bicycle=designated (chemin autorisé) : acceptable
                    WHEN e.highway = 'path' AND e.bicycle = 'designated' THEN 2.0
                    WHEN e.highway = 'track' THEN 2.5
                    WHEN e.highway = 'busway' THEN 1.5
                    
                    ELSE 2.0
                END AS hierarchy_f,
                
                -- 2. SURFACE (pour un 25mm route)
                CASE
                    WHEN e.surface IN ('asphalt', 'paved', 'concrete') THEN 1.0
                    WHEN e.surface = 'paving_stones' THEN 1.4
                    WHEN e.surface = 'sett' THEN 1.8
                    WHEN e.surface IN ('cobblestone', 'unhewn_cobblestone') THEN 3.0
                    WHEN e.surface = 'fine_gravel' THEN 2.0
                    WHEN e.surface IN ('gravel', 'compacted') THEN 2.5
                    WHEN e.surface IN ('unpaved', 'ground', 'dirt', 'earth') THEN 4.0
                    WHEN e.surface = 'grass' THEN 5.0
                    WHEN e.surface = 'sand' THEN 8.0
                    WHEN e.surface IS NULL THEN 1.0
                    ELSE 1.5
                END AS surface_f,
                
                -- 3. BNAC CONTEXTUEL (sur axes dangereux, la piste sauve)
                CASE
                    WHEN e.has_bnac AND e.highway IN ('primary', 'primary_link') THEN 0.7
                    WHEN e.has_bnac AND e.highway IN ('secondary', 'secondary_link') THEN 0.85
                    WHEN e.has_bnac AND e.highway IN ('tertiary', 'unclassified') THEN 1.0
                    WHEN e.has_bnac AND e.highway = 'residential' THEN 1.05
                    WHEN e.has_bnac AND e.highway IN ('path', 'footway') THEN 1.2
                    ELSE 1.0
                END AS bnac_f,
                
                -- 4. RESIDENTIEL : village vs ville (node_density)
                -- Village (faible densité) : bonus
                -- Ville dense : malus fort
                CASE
                    WHEN e.highway IN ('residential', 'living_street') THEN
                        CASE
                            WHEN e.node_density IS NULL THEN 1.0
                            WHEN e.node_density < 50 THEN 0.9    -- village : sympa
                            WHEN e.node_density < 150 THEN 1.0   -- périurbain : neutre
                            WHEN e.node_density < 300 THEN 1.3   -- urbain : malus
                            ELSE 1.6                             -- centre-ville dense
                        END
                    ELSE 1.0
                END AS residential_f,
                
                -- 5. ZONES VERTES : bonus proportionnel à la fraction verte
                -- greenery_ratio = [0, 1], bonus jusqu'à 0.85 si totalement en vert
                CASE
                    WHEN e.greenery_ratio IS NULL THEN 1.0
                    ELSE 1.0 - 0.15 * COALESCE(e.greenery_ratio, 0)
                END AS greenery_f
                
            FROM osm_edges e
            WHERE e.is_routable = TRUE
        )
        UPDATE edge_scores es
        SET cost_factor_club = GREATEST(1.0,
            b.hierarchy_f * b.surface_f * b.bnac_f * b.residential_f * b.greenery_f
        ),
            score_osm_club = 1.0 / (1.0 + (b.hierarchy_f * b.surface_f * b.bnac_f * b.residential_f * b.greenery_f - 1.0))
        FROM base b
        WHERE es.edge_id = b.edge_id;
    """))

print("✓ cost_factor_club calculé")
````

````python
# --- Même logique pour le profil SOLO_CASUAL ---
# Différences : pistes cyclables encouragées, résidentiel dense moins punitif,
# secondary plus tolérées car moins de compromis sur la vitesse

print("Calcul cost_factor_solo...")

with engine.begin() as conn:
    conn.execute(text("""
        WITH base AS (
            SELECT e.edge_id,
                -- 1. HIERARCHIE : plus tolérante sur les axes lents
                CASE
                    WHEN e.highway IN ('tertiary', 'unclassified', 'tertiary_link') THEN 1.0
                    WHEN e.highway = 'secondary' AND (e.maxspeed IS NULL OR e.maxspeed < 70) THEN 1.2
                    WHEN e.highway = 'secondary' AND e.maxspeed BETWEEN 70 AND 89 THEN 2.0
                    WHEN e.highway = 'secondary_link' THEN 1.3
                    WHEN e.highway = 'primary' AND e.maxspeed < 70 THEN 1.8
                    WHEN e.highway = 'primary' AND e.maxspeed BETWEEN 70 AND 89 THEN 4.0
                    WHEN e.highway = 'primary_link' THEN 2.5
                    WHEN e.highway = 'residential' THEN 1.1
                    WHEN e.highway = 'living_street' THEN 1.2
                    WHEN e.highway = 'cycleway' THEN 0.85    -- bonus solo
                    WHEN e.highway = 'service' THEN 2.0
                    WHEN e.highway = 'path' AND e.bicycle = 'designated' THEN 1.5
                    WHEN e.highway = 'track' THEN 2.0
                    WHEN e.highway = 'busway' THEN 1.3
                    ELSE 1.8
                END AS hierarchy_f,
                
                -- 2. SURFACE : un peu moins dur sur gravel/compacted (vélo plus gros accepté)
                CASE
                    WHEN e.surface IN ('asphalt', 'paved', 'concrete') THEN 1.0
                    WHEN e.surface = 'paving_stones' THEN 1.2
                    WHEN e.surface = 'sett' THEN 1.5
                    WHEN e.surface IN ('cobblestone', 'unhewn_cobblestone') THEN 2.5
                    WHEN e.surface = 'fine_gravel' THEN 1.3
                    WHEN e.surface IN ('gravel', 'compacted') THEN 1.5
                    WHEN e.surface IN ('unpaved', 'ground', 'dirt', 'earth') THEN 2.5
                    WHEN e.surface = 'grass' THEN 3.5
                    WHEN e.surface = 'sand' THEN 6.0
                    WHEN e.surface IS NULL THEN 1.0
                    ELSE 1.3
                END AS surface_f,
                
                -- 3. BNAC : plus de bonus (solo apprécie plus la sécurité)
                CASE
                    WHEN e.has_bnac AND e.highway IN ('primary', 'primary_link') THEN 0.6
                    WHEN e.has_bnac AND e.highway IN ('secondary', 'secondary_link') THEN 0.8
                    WHEN e.has_bnac AND e.highway IN ('tertiary', 'unclassified') THEN 0.95
                    WHEN e.has_bnac AND e.highway = 'residential' THEN 0.95
                    WHEN e.has_bnac AND e.highway IN ('path', 'footway') THEN 1.0
                    ELSE 1.0
                END AS bnac_f,
                
                -- 4. RESIDENTIEL : très peu de malus pour zones denses (acceptable en solo)
                CASE
                    WHEN e.highway IN ('residential', 'living_street') THEN
                        CASE
                            WHEN e.node_density IS NULL THEN 1.0
                            WHEN e.node_density < 50 THEN 0.9
                            WHEN e.node_density < 150 THEN 1.0
                            WHEN e.node_density < 300 THEN 1.15
                            ELSE 1.3
                        END
                    ELSE 1.0
                END AS residential_f,
                
                -- 5. ZONES VERTES : même bonus
                CASE
                    WHEN e.greenery_ratio IS NULL THEN 1.0
                    ELSE 1.0 - 0.15 * COALESCE(e.greenery_ratio, 0)
                END AS greenery_f
                
            FROM osm_edges e
            WHERE e.is_routable = TRUE
        )
        UPDATE edge_scores es
        SET cost_factor_solo = GREATEST(1.0,
            b.hierarchy_f * b.surface_f * b.bnac_f * b.residential_f * b.greenery_f
        ),
            score_osm_solo = 1.0 / (1.0 + (b.hierarchy_f * b.surface_f * b.bnac_f * b.residential_f * b.greenery_f - 1.0))
        FROM base b
        WHERE es.edge_id = b.edge_id;
    """))

print("✓ cost_factor_solo calculé")
````

### Statistiques sur les cost_factors

On vérifie que la distribution est saine.

````python
with engine.connect() as conn:
    stats = pd.read_sql(text("""
        SELECT
            e.highway,
            COUNT(*) AS n,
            ROUND(AVG(es.cost_factor_club)::numeric, 2) AS club_mean,
            ROUND(MIN(es.cost_factor_club)::numeric, 2) AS club_min,
            ROUND(MAX(es.cost_factor_club)::numeric, 2) AS club_max,
            ROUND(AVG(es.cost_factor_solo)::numeric, 2) AS solo_mean
        FROM osm_edges e
        JOIN edge_scores es ON es.edge_id = e.edge_id
        WHERE e.is_routable = TRUE
          AND e.highway IN ('tertiary', 'unclassified', 'secondary', 'primary',
                            'residential', 'cycleway', 'path', 'service', 'living_street')
        GROUP BY e.highway
        ORDER BY club_mean;
    """), conn)
print("Cost factors par type de route :")
print(stats.to_string(index=False))
````

````python
# Contrôle : chercher si des arêtes ont un cost_factor aberrant (>15)
with engine.connect() as conn:
    aber = pd.read_sql(text("""
        SELECT
            e.highway, e.maxspeed, e.surface,
            COUNT(*) AS n,
            ROUND(AVG(es.cost_factor_club)::numeric, 2) AS mean_cost
        FROM osm_edges e
        JOIN edge_scores es ON es.edge_id = e.edge_id
        WHERE es.cost_factor_club > 15 AND e.is_routable = TRUE
        GROUP BY e.highway, e.maxspeed, e.surface
        ORDER BY n DESC LIMIT 10;
    """), conn)
print("Arêtes à cost_factor >15 (à vérifier) :")
print(aber.to_string(index=False))
````

## 5. Score final composé

On combine `score_osm` (inverse du cost_factor) + `score_club` (heatmap
brute) en un score final entre 0 et 1.

$$\text{score\_final} = 0.55 \cdot \text{score\_club} + 0.45 \cdot \text{score\_osm}$$

Ces poids sont provisoires. Au Notebook 5, on remplacera `score_club` par
`score_club_gen` (PU learning) et on apprendra les poids optimaux.

````python
W_CLUB = 0.55
W_OSM  = 0.45

with engine.begin() as conn:
    # Ajout colonnes si besoin
    conn.execute(text("""
        ALTER TABLE edge_scores
        ADD COLUMN IF NOT EXISTS score_final_club DOUBLE PRECISION DEFAULT 0;
        ALTER TABLE edge_scores
        ADD COLUMN IF NOT EXISTS score_final_solo DOUBLE PRECISION DEFAULT 0;
    """))
    
    conn.execute(text(f"""
        UPDATE edge_scores
        SET score_final_club = {W_CLUB} * score_club + {W_OSM} * score_osm_club,
            score_final_solo = {W_CLUB} * score_club + {W_OSM} * score_osm_solo,
            updated_at = NOW();
    """))

with engine.connect() as conn:
    s = pd.read_sql(text("""
        SELECT
            ROUND(AVG(score_final_club)::numeric, 3) AS club_mean,
            ROUND(AVG(score_final_solo)::numeric, 3) AS solo_mean,
            COUNT(*) FILTER (WHERE score_final_club >= 0.5) AS club_premium,
            COUNT(*) FILTER (WHERE score_final_solo >= 0.5) AS solo_premium
        FROM edge_scores es JOIN osm_edges e ON e.edge_id = es.edge_id
        WHERE e.is_routable;
    """), conn)
print(s.to_string(index=False))
````

## 6. Visualisation

On affiche la heatmap de score_final pour les deux profils côte à côte. On
doit voir des différences sur les axes avec piste cyclable (avantage solo)
et les zones résidentielles denses (malus club plus fort).

````python
import folium
import branca.colormap as cm
from shapely import wkt

with engine.connect() as conn:
    top = pd.read_sql(text("""
        SELECT e.highway, e.has_bnac, e.node_density, e.greenery_ratio,
               es.score_final_club, es.score_final_solo,
               ST_AsText(ST_Simplify(e.geom, 0.0002)) AS wkt
        FROM osm_edges e JOIN edge_scores es ON es.edge_id = e.edge_id
        WHERE e.is_routable
          AND (es.score_final_club > 0.3 OR es.score_final_solo > 0.3)
        ORDER BY (es.score_final_club + es.score_final_solo) DESC
        LIMIT 15000;
    """), conn)
print(f"{len(top):,} arêtes à afficher")
````

````python
# Carte simple : profil club_road
m = folium.Map(location=[48.75, 2.3], zoom_start=10, tiles="cartodbpositron")
colormap = cm.linear.YlOrRd_09.scale(0.3, 1.0)

for _, r in top.iterrows():
    try:
        g = wkt.loads(r["wkt"])
        coords = [(y, x) for x, y in g.coords]
        folium.PolyLine(
            coords, color=colormap(r["score_final_club"]),
            weight=2, opacity=0.7,
            tooltip=f"{r['highway']} | club={r['score_final_club']:.2f} solo={r['score_final_solo']:.2f}"
        ).add_to(m)
    except Exception:
        pass

m.add_child(colormap)
m
````
