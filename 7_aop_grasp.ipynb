# Notebook 7 — AOP GRASP rigoureux (perspective)

## Référence

Verbeeck, C., Vansteenwegen, P., Aghezzaf, E. (2014).
*The Arc Orienteering Problem.* 
Transportation Research Part E: Logistics and Transportation Review, 68, 64-78.

## Problème mathématique

Soit G = (V, E) un graphe orienté. Chaque arête e ∈ E a :
- Un coût (longueur, ou cost_factor × longueur)
- Un score (notre `score_final_club`)

Étant donnés A (départ), B (arrivée), et un budget T (temps/distance max),
trouver le chemin de A à B qui :
- Maximise la **somme** des scores des arêtes parcourues
- Respecte le budget total ≤ T
- Sans répéter d'arête (problème NP-difficile)

## Algorithme GRASP

GRASP = Greedy Randomized Adaptive Search Procedure. Deux phases répétées :

1. **Construction gloutonne randomisée** : à chaque étape, choix parmi les meilleurs
   k candidats (RCL = Restricted Candidate List)
2. **Local search** : amélioration par échanges locaux

Pour AOP cyclique : 50 itérations, RCL = top 10%, local search par 2-opt.

- Temps de calcul : 30-120s par A→B

## Note méthodologique pour le mémoire

Le Notebook 6 (super-arêtes par sous-tronçons club) est une **heuristique simple**
de l'AOP. Pour passer à un algo rigoureux, il faut :

1. Construire une "table des distances" entre tous les nœuds importants (~5000)
2. Réécrire le problème en ILP (Integer Linear Programming)
3. Soit résoudre avec Gurobi/CPLEX (pour optimal exact)
4. Soit utiliser GRASP/ALNS pour solution approchée rapide

C'est un projet de stage M2 à part entière. Bien défendable dans la perspective.
